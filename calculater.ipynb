```python
answer = 0
def add(num1, num2):
    return num1 + num2

def subtract(num1, num2):
    return num1 - num2

def multiply(num1, num2):
    return num1 * num2 

def divide(num1, num2):
    if num2 == 0:
        return(print("Error! cannot divide by zero"))
    else:
        return num1 / num2

            
while True:
    num1 = input("Enter first number: ")
    num2 = input("Enter second number: ")
    print(num1, num2)
    
    if num1.isdigit():
        num1 = int(num1)
    else:
        print("Error! please enter a integer")
        continue

    if num2.isdigit():
        num2 = int(num2)
    else:
        print("Error! please enter a integer")
        continue

    operator = int(input("Select a Operator:"
                " 1. Addition "
                " 2. Subtraction "
                " 3. Multiplication "
                " 4. Division "
                ))

    if operator == 1: 
        print(num1, "+", num2, "=", add(num1, num2))
    elif operator == 2:
        print(num1, "-", num2, "=", subtract(num1, num2))
    elif operator == 3:
        print(num1, "*", num2, "=", multiply(num1, num2))
    else:
        print(num1, "/", num2, "=", divide(num1, num2))

    loop_again = input("Would you like to calculate another answer? (yes/no): ")
    if loop_again == "no":
          break
    else:
        continue

print("Thank you for playing :)") 
```
